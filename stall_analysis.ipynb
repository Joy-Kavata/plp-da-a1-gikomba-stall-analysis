Gikomba Market Stall Performance Analysis

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

Task 1: Overall Daily Sales Metrics

```python
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
max_sales = max(daily_sales)
min_sales = min(daily_sales)

max_index = daily_sales.index(max_sales)
min_index = daily_sales.index(min_sales)

best_day_name = days[max_index]
best_week = (max_index // 7) + 1

worst_day_name = days[min_index]
worst_week = (min_index // 7) + 1

print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {avg_sales:,.2f}")
print(f"Highest Sales: KES {max_sales:,} ({best_day_name}, Week {best_week})")
print(f"Lowest Sales: KES {min_sales:,} ({worst_day_name}, Week {worst_week})")
```

Task 2: Week-on-Week Growth

```python
def week_totals(sales):
    """
    Takes a 14-day sales list and returns a tuple of (week_1_total, week_2_total)
    using Python slicing.
    """
    week1_total = sum(sales[:7])
    week2_total = sum(sales[7:])
    return week1_total, week2_total

wk1_tot, wk2_tot = week_totals(daily_sales)
wow_growth = ((wk2_tot - wk1_tot) / wk1_tot) * 100

print(f"Week 1 Total: KES {wk1_tot:,}")
print(f"Week 2 Total: KES {wk2_tot:,}")
print(f"Week-on-Week Growth: {wow_growth:.2f}%")
```

Task 3: Day Classification

```python
def classify_day(amount, average):
    """
    Classifies a daily sales amount relative to the average:
    - 'Strong': > 20% above average
    - 'Weak': > 20% below average
    - 'Normal': Within 20% of average
    """
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

classification_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    cat = classify_day(amount, avg_sales)
    classification_counts[cat] += 1

print("Day Classifications Count:")
for category, count in classification_counts.items():
    print(f"  {category}: {count} days")
```

Task 4: Weekdat vs. Weekend Analysis

```python
weekday_sales = []
weekend_sales = []

for i, amount in enumerate(daily_sales):
    # Indices 5, 6, 12, 13 correspond to Sat and Sun
    if days[i] in ["Sat", "Sun"]:
        weekend_sales.append(amount)
    else:
        weekday_sales.append(amount)

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
diff_percentage = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(f"Average Weekday Sales: KES {avg_weekday:,.2f}")
print(f"Average Weekend Sales: KES {avg_weekend:,.2f}")
print(f"Weekend vs Weekday Difference: {diff_percentage:.2f}% higher on weekends")
```

Task 5 - Stock & Reorder Analysis

```python
def get_reorder_products(stock_dict):
    """
    Returns a list of product names whose current quantity is at or below
    their reorder level.
    """
    reorder_list = []
    for product, info in stock_dict.items():
        if info["qty"] <= info["reorder"]:
            reorder_list.append(product)
    return reorder_list

items_to_reorder = get_reorder_products(stock)

print("Products needing restock:", items_to_reorder)
print("\nGross Margins by Product:")
for product, info in stock.items():
    margin = ((info["price"] - info["cost"]) / info["price"]) * 100
    print(f"  {product}: {margin:.2f}% (Cost: KES {info['cost']}, Price: KES {info['price']})")
```

Task 6 - Formatted Stall Performance Report

```python
print("=" * 55)
print(f"{'STALL PERFORMANCE REPORT':^55}")
print("=" * 55)
print(f"Total 14-Day Revenue:          KES {total_sales:>10,}")
print(f"Average Daily Sales:           KES {avg_sales:>10,.2f}")
print(f"Week 1 Revenue:                KES {wk1_tot:>10,}")
print(f"Week 2 Revenue:                KES {wk2_tot:>10,}")
print(f"Week-on-Week Growth:           {wow_growth:>13.2f}%")
print("-" * 55)
print(f"Average Weekday Sales:         KES {avg_weekday:>10,.2f}")
print(f"Average Weekend Sales:         KES {avg_weekend:>10,.2f}")
print("-" * 55)
print(f"Products Needing Restock:      {', '.join(items_to_reorder)}")
print("=" * 55)
```
